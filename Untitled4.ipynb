```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATE, CATEGORY, REGION, VALUE, COST = "date", "category", "region", "sales", "cost"
sns.set_theme(style="whitegrid", palette="deep", font_scale=1.05)


def make_sample_data(seed=42):
    rng = np.random.default_rng(seed)
    dates = pd.date_range("2024-01-01", "2025-12-01", freq="MS")
    rows = []
    for d in dates:
        season = 1 + 0.25 * np.sin((d.month - 3) / 12 * 2 * np.pi) + (0.2 if d.month == 12 else 0)
        growth = 1 + 0.015 * ((d.year - 2024) * 12 + d.month)
        for region, r_w in [("North", 1.0), ("South", 0.8), ("East", 1.2), ("West", 0.9)]:
            for cat, c_w, margin in [("Electronics", 1.6, 0.18), ("Clothing", 1.0, 0.35), ("Home", 0.7, 0.28)]:
                sales = 10000 * r_w * c_w * season * growth * rng.normal(1, 0.07)
                rows.append([d, cat, region, round(sales, 2), round(sales * (1 - margin) * rng.normal(1, 0.03), 2)])
    return pd.DataFrame(rows, columns=[DATE, CATEGORY, REGION, VALUE, COST])


df = make_sample_data()
df = df.dropna(subset=[DATE, VALUE]).copy()
df["profit"] = df[VALUE] - df[COST]
df["month"] = df[DATE].dt.to_period("M").dt.to_timestamp()
print(df.head())


def finish(fig, ax, title, subtitle):
    ax.set_title(title, loc="left", fontsize=14, fontweight="bold")
    fig.text(0.125, 0.92, subtitle, fontsize=10, color="gray")
    sns.despine()
    plt.show()


# Chart 1: trend over time
monthly = df.groupby("month")[VALUE].sum()
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(monthly.index, monthly.values, marker="o", linewidth=2)
ax.set_xlabel("Month"); ax.set_ylabel("Total sales")
finish(fig, ax, "Sales climb each year and peak in December", "Monthly total sales")

# Chart 2: comparison
fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=df, x=REGION, y=VALUE, hue=CATEGORY, estimator="sum", errorbar=None, ax=ax)
ax.set_xlabel("Region"); ax.set_ylabel("Total sales")
finish(fig, ax, "East leads every category", "Total sales by region and category")

# Chart 3: distribution
fig, ax = plt.subplots(figsize=(10, 5))
sns.boxplot(data=df, x=CATEGORY, y="profit", ax=ax)
ax.set_xlabel("Category"); ax.set_ylabel("Profit per region-month")
finish(fig, ax, "Electronics has the widest profit swings", "Distribution of profit by category")

# Chart 4: relationship
fig, ax = plt.subplots(figsize=(10, 5))
sns.scatterplot(data=df, x=VALUE, y="profit", hue=CATEGORY, alpha=0.7, ax=ax)
ax.set_xlabel("Sales"); ax.set_ylabel("Profit")
finish(fig, ax, "Higher sales don't mean higher profit", "Sales vs profit by category")

# Chart 5: heatmap
pivot = df.assign(m=df[DATE].dt.month).pivot_table(index=REGION, columns="m", values=VALUE, aggfunc="sum")
fig, ax = plt.subplots(figsize=(11, 4))
sns.heatmap(pivot, cmap="YlGnBu", annot=True, fmt=".0f", cbar_kws={"label": "Sales"}, ax=ax)
ax.set_xlabel("Month of year"); ax.set_ylabel("")
finish(fig, ax, "Seasonality is consistent across regions", "Sales by region and calendar month")

# Chart 6: margin by category
margin = (df.groupby(CATEGORY)["profit"].sum() / df.groupby(CATEGORY)[VALUE].sum() * 100).sort_values()
fig, ax = plt.subplots(figsize=(8, 4))
ax.barh(margin.index, margin.values)
for i, v in enumerate(margin.values):
    ax.text(v + 0.4, i, f"{v:.1f}%", va="center")
ax.set_xlabel("Profit margin (%)")
finish(fig, ax, "Clothing is the most profitable category", "Profit margin by category")

print("Best region:", df.groupby(REGION)[VALUE].sum().idxmax())
print("Top margin category:", margin.idxmax())
print("Peak month:", monthly.idxmax().strftime("%B %Y"))
```
